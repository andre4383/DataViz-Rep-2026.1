```python
import pandas as pd
import numpy as np
```

```python
# 1. Carregar os dados de emprego
df_emprego = pd.read_csv('dados/emprego.csv', sep=';')
```

```python
# 2. Limpeza e Tratamento
# Converter período para datetime
df_emprego['periodo'] = pd.to_datetime(df_emprego['periodo'])

# Substituir os zeros (gap da pandemia) por NaN e remover essas linhas
df_emprego.replace(0.0, np.nan, inplace=True)
df_emprego_clean = df_emprego.dropna().reset_index(drop=True)
```

```python
# 3. Exibição da tabela limpa
print("--- Amostra dos Dados de Emprego (Limpos) ---")
display(df_emprego_clean.head())
print(f"Total de linhas válidas após limpeza: {len(df_emprego_clean)}\n")
```

```python
# 4. Geração de Métricas para a Documentação
media_total = df_emprego_clean['total'].mean()
media_homens = df_emprego_clean['homens'].mean()
media_mulheres = df_emprego_clean['mulheres'].mean()

# Encontrando os picos e vales do total
pico_emprego = df_emprego_clean.loc[df_emprego_clean['total'].idxmax()]
vale_emprego = df_emprego_clean.loc[df_emprego_clean['total'].idxmin()]

print("--- Métricas-Chave para Adicionar na Documentação ---")
print(f"A. Taxa Média Histórica de Emprego (Geral): {media_total:.2f}%")
print(f"B. Taxa Média Histórica - Homens: {media_homens:.2f}%")
print(f"C. Taxa Média Histórica - Mulheres: {media_mulheres:.2f}%")
```

```python
gap_genero = media_homens - media_mulheres
print(f"D. Gap (Desigualdade) Médio de Gênero: {gap_genero:.2f} pontos percentuais")
```

```python
print(f"\nE. Período com MAIOR taxa de emprego geral: {pico_emprego['periodo'].strftime('%m/%Y')} ({pico_emprego['total']}%)")
print(f"F. Período com MENOR taxa de emprego geral: {vale_emprego['periodo'].strftime('%m/%Y')} ({vale_emprego['total']}%)")
```
