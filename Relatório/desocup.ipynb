```python
import pandas as pd
import numpy as np
```

```python
# 1. Carregar os dados (ajuste o caminho se necessário para dentro da pasta /dados)
df_desocupacao = pd.read_csv('dados/desocupacao.csv', sep=';')
df_emprego = pd.read_csv('dados/emprego.csv', sep=';')
df_informalidade = pd.read_csv('dados/informalidade.csv', sep=';')
```

```python
# 2. Função de limpeza (Pipeline de Tratamento)
def limpar_dados(df):
    # Converte a coluna 'periodo' para datetime para facilitar análises temporais
    df['periodo'] = pd.to_datetime(df['periodo'])
    
    # Substitui 0.0 por NaN para evitar distorção nas médias
    df.replace(0.0, np.nan, inplace=True)
    
    # Remove as linhas que contêm NaN
    df_limpo = df.dropna().reset_index(drop=True)
    
    return df_limpo
```

```python
# 3. Aplicar a limpeza aos DataFrames
df_desocupacao_clean = limpar_dados(df_desocupacao)
df_emprego_clean = limpar_dados(df_emprego)
df_informalidade_clean = limpar_dados(df_informalidade)
```

```python
# 4. Exibir os resultados e confirmar que os zeros saíram
print("--- Dados de Desocupação Limpos (Amostra) ---")
display(df_desocupacao_clean.head())
```
