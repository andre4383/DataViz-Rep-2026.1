```python
import pandas as pd
import numpy as np
```

```python
# 1. Carregar os dados de informalidade
df_informalidade = pd.read_csv('dados/informalidade.csv', sep=';')  
```

```python
# 2. Limpeza e Tratamento
# Converter período para datetime
df_informalidade['periodo'] = pd.to_datetime(df_informalidade['periodo'])

# Substituir os zeros (gap inicial de 2012-2015 e gap da pandemia) por NaN e remover essas linhas
df_informalidade.replace(0.0, np.nan, inplace=True)
df_informalidade_clean = df_informalidade.dropna().reset_index(drop=True)
```

```python
# 3. Exibição da tabela limpa
print("--- Amostra dos Dados de Informalidade (Limpos) ---")
display(df_informalidade_clean.head())
print(f"Total de linhas válidas após limpeza: {len(df_informalidade_clean)}\n")
```

```python
# 4. Geração de Métricas para a Documentação
media_total = df_informalidade_clean['total'].mean()
media_homens = df_informalidade_clean['homens'].mean()
media_mulheres = df_informalidade_clean['mulheres'].mean()

# Encontrando os picos e vales do total
pico_informalidade = df_informalidade_clean.loc[df_informalidade_clean['total'].idxmax()]
vale_informalidade = df_informalidade_clean.loc[df_informalidade_clean['total'].idxmin()]
```

```python
print("--- Métricas-Chave para Adicionar na Documentação ---")
print(f"A. Taxa Média Histórica de Informalidade (Geral): {media_total:.2f}%")
print(f"B. Taxa Média Histórica - Homens: {media_homens:.2f}%")
print(f"C. Taxa Média Histórica - Mulheres: {media_mulheres:.2f}%")
```

```python
# Calculando a diferença (neste caso, homens têm taxas maiores)
gap_genero = media_homens - media_mulheres
print(f"D. Diferença Média de Gênero (Homens > Mulheres): {gap_genero:.2f} pontos percentuais")
```

```python
print(f"\nE. Período com MAIOR taxa de informalidade geral: {pico_informalidade['periodo'].strftime('%m/%Y')} ({pico_informalidade['total']}%)")
print(f"F. Período com MENOR taxa de informalidade geral: {vale_informalidade['periodo'].strftime('%m/%Y')} ({vale_informalidade['total']}%)")
```
